```python
def h(n):
 if n == 1:
    return 1
 else:
  return n * h(n-1)
print(h(5))
```

```python
def f(n):
    if n == 1:
        return 1
    elif n == 2:
        return 2
    else:
        return f(n-1) + f(n-2)

print(f"總共有{f(10)}種方式")
```
